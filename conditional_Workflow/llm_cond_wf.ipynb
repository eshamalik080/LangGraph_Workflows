```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict, Literal
from dotenv import load_dotenv
from pydantic import BaseModel, Field
```

```python
load_dotenv()

model = ChatGroq(model="openai/gpt-oss-120b", temperature=0)
```

```python
class SentimentSchema(BaseModel):
    sentiment: Literal["positive", "negative", "neutral"] = Field(description="The sentiment of the text.")
```

```python
structured_model = model.with_structured_output(SentimentSchema)
```

```python
prompt = """ What is the sentiment of the following review - The software too bad """
structured_output = structured_model(prompt)
```
